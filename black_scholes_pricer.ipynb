```python
import numpy as np
from scipy.stats import norm
```

```python
import yfinance as yf

# Pull real NVDA data
stock = yf.download("NVDA", start="2020-01-01", end="2025-01-01", progress=False)

# Calculate real inputs from historical data
close_prices = stock["Close"]["NVDA"]
daily_returns = close_prices.pct_change().dropna()

S = float(close_prices.iloc[-1])   # Last known NVDA price
sigma = float(daily_returns.std() * np.sqrt(252))  # Annualized volatility
K = round(S * 1.05, 2)             # Strike price 5% above current price
T = 1.0                             # 1 year to expiration
r = 0.05                            # Risk free rate

print(f"Stock price:   ${S:.2f}")
print(f"Strike price:  ${K:.2f}")
print(f"Volatility:    {sigma*100:.1f}%")
print(f"Time:          {T} year")
print(f"Interest rate: {r*100}%")
```

```python
# Step 1 - calculate d1 and d2 (intermediate values the formula needs)
d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
d2 = d1 - sigma * np.sqrt(T)

print(f"d1: {d1:.4f}")
print(f"d2: {d2:.4f}")
```

```python
# Step 2 - calculate the call option price
call_price = S * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)

print(f"NVDA Call Option Price: ${call_price:.2f}")
```

```python
# Step 3 - calculate the put option price
put_price = K * np.exp(-r * T) * norm.cdf(-d2) - S * norm.cdf(-d1)

print(f"NVDA Call Option Price: ${call_price:.2f}")
print(f"NVDA Put Option Price:  ${put_price:.2f}")
```

```python
# Final summary
print("=" * 40)
print("   NVDA Options Pricer - Black-Scholes")
print("=" * 40)
print(f"Stock Price:      ${S:.2f}")
print(f"Strike Price:     ${K:.2f}")
print(f"Volatility:       {sigma*100:.1f}%")
print(f"Time to Expiry:   {T} year")
print(f"Risk Free Rate:   {r*100}%")
print("=" * 40)
print(f"CALL Option:      ${call_price:.2f}")
print(f"PUT Option:       ${put_price:.2f}")
print("=" * 40)
```
